## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*


Task Type: Classification.

Why: The goal is to categorize aging, low-traffic pages into two distinct buckets: "Safe to Prune" (delete/redirect) or "Risky to Prune" (keep/update). We are predicting a binary outcome to give the site manager a clear go/no-go decision.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target: is_safe_to_prune.

Origin: This is a proxy label because it is based on a defined historical rule rather than a strict future-observed outcome. We will define "safe" as having near-zero impressions and clicks over the trailing 90 days, combined with a high content_age_days.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: Precision (on the "Safe to Prune" class).

What means "good": We must define what number constitutes a good result. A false positive here is dangerous—advising a manager to delete a page that was actually valuable. Therefore, a "good" model must achieve exceptionally high Precision (e.g., 95%+), ensuring that when it flags a page as safe to delete, it is almost never wrong.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML beats an if-statement: Relying on simple if/else business rules is too messy or complex for this problem. A rigid rule (e.g., "IF impressions = 0 THEN delete") might accidentally flag highly seasonal pages or brand-new content. ML earns its keep by weighing the full context—age, engagement, and visibility trends—to detect nuanced risks that a basic SQL rule would miss.
