## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Task Type: Ranking / Scoring.
Why: The goal is not just to classify if a page is declining, but to assign a priority score so we can rank candidates from highest opportunity to lowest. The content team needs a ranked queue (e.g., "Which 50 pages should we fix first?") rather than an unranked list.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target: is_declining_label, which is defined as trend_direction == "down".
Origin: This is a proxy label derived from a rule calculating recent trailing trends. It is not a pure future-observed outcome (like "did traffic drop in the 30 days after prediction"), but it serves as a measurable stand-in for underperformance that warrants human review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: Precision@50.
What means "good": Precision@50 measures what percentage of the top 50 ranked pages actually turned out to be positive for our decline label. "Good" means significantly beating the static baseline rules. Since the baseline achieves a Precision@50 of 0.240 (24%), a "good" model score would be anything consistently higher than that, proving the model surfaces better candidates.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML beats an if-statement: A simple fixed rule (e.g., "IF age > 180 days AND position > 10 THEN refresh") is too rigid. The pattern of content decay is messy: it involves tangled interactions between age, search volume, click-through rates, and engagement metrics. ML earns its keep here by weighing dozens of observable signals simultaneously to find non-linear patterns that hand-written SQL logic would miss, resulting in a much more accurate review queue.
