## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

```python
Lane: Refresh / Content Opportunity Scoring.
Why: Content teams have limited capacity and need to prioritize which underperforming pages to update first. This lane focuses on building a ranked queue of candidate pages based on observable search and engagement signals.
It provides actionable decision support rather than just a theoretical model, directly addressing the business need to allocate editor time effectively.
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
Decision: Which specific pages should the content team review first for a potential refresh or update?
Action: A human editor will review the top-ranked pages in the queue and execute on-page changes (like updating metadata, expanding content, or pruning).
Cost of a wrong call: Wasting expensive editor hours on pages that are merely experiencing seasonal shifts or have no real search demand, while missing true opportunities where a refresh could actually recover lost traffic.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

github_url = 'https://raw.githubusercontent.com/Netrahoni/FlyRankAi-Intern-work-Files/main/data/raw/content_refresh_anonymized.csv'

df = pd.read_csv(github_url)

total_rows = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
pages_with_demand = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] > 500)])

print(f"Total content items in starter dataset: {total_rows:,}")
print(f"Pages showing a downward trend: {declining_pages:,}")
print(f"Declining pages with significant demand (>500 impressions): {pages_with_demand:,}")
print("\nConclusion: With thousands of declining pages showing real demand, a human cannot manually review all of them. This justifies building a ranked scoring model to prioritize the queue.")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
What I can claim: I can claim that my analysis identifies observable signals (like age, search volume, and CTR) associated with traffic movement. I can claim that my model provides directional, decision-support by ranking
candidates more effectively than a static baseline rule.
What I cannot claim: I cannot claim that performing a refresh will cause a page to recover, as causal proof requires an experiment rather than observational data. Furthermore, I cannot claim to have "proven a Google algorithm factor" or
reverse-engineered search rankings.
```
