## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

A ranked queue of underperforming pages, sorted by the highest potential click recovery.
The Action: An SEO specialist or content editor reviews the page and rewrites the title tag and meta description to better match search intent and entice clicks.
Reason Codes: Each page in the queue will be tagged with a reason code so the editor knows why it was flagged:

high_impression_low_ctr: Page has massive search volume but users are scrolling past it.

page_1_underperformer: Page is ranking in the top 10 positions but has a CTR significantly below the baseline for its exact spot.

stale_metadata: High anomaly score combined with content that hasn't been updated in over a year.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

Intended Use: This playbook is a decision-support tool. It is designed to be pulled weekly by the editorial team to prioritize their rewrite queue, ensuring they spend their limited hours on the pages with the highest mathematical probability of recovering lost traffic.
Cost/Value: The value is maximizing editorial ROI by focusing on proven search demand. The cost of a wrong call (a false positive) is 15-30 minutes of an editor's time rewriting a title that didn't actually need it.
Limits: This model identifies opportunities, not guarantees. It cannot predict if a low CTR is caused by external factors like Google introducing a new AI Overview or a rich snippet above our link.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

Human-Review Rules: A human must look at the actual Search Engine Results Page (SERP) before rewriting. If the search intent is purely informational (a "zero-click" search where the user gets the answer without clicking), the page should be skipped.
What should NOT be automated (The No-Go List):

Automated Rewriting: Do NOT automatically pass these URLs to an LLM to rewrite and publish the titles without a human in the loop (high risk of brand voice damage or hallucination).

Recent Updates: Do NOT flag pages that have had their metadata updated in the last 30 days. They need time to settle in the search index before being evaluated again.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

Copy this into the Markdown cell for Section 4:
Light Monitoring: Track the 30-day post-rewrite CTR of the pages the team actually updates. If the aggregated CTR of updated pages does not improve over a 90-day quarter, the intervention itself (the rewrites) is failing.
Retrain Triggers:

Data Shift: Retrain the model if a major Google core update drastically changes baseline CTR expectations across the site.

Performance Drop: Retrain if the model's Precision drops below our established baseline (meaning it starts flagging too many normal pages as anomalies).

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

```python
import pandas as pd
import os
import matplotlib.pyplot as plt

os.makedirs('../../work/outputs', exist_ok=True)
os.makedirs('../../work/figures', exist_ok=True)

data = {
    'content_id': ['page_A', 'page_B', 'page_C', 'page_D'],
    'is_ctr_anomaly': [1, 0, 1, 0],
    'anomaly_probability': [0.89, 0.21, 0.75, 0.15],
    'impressions_90d': [15000, 200, 12000, 5000],
    'reason_code': ['high_impression_low_ctr', 'none', 'page_1_underperformer', 'none']
}
df_scored = pd.DataFrame(data)

df_queue = df_scored[df_scored['is_ctr_anomaly'] == 1].sort_values(by='anomaly_probability', ascending=False)

queue_filename = '../../work/outputs/action_queue.csv'
df_queue.to_csv(queue_filename, index=False)
print(f"✅ Ranked action queue successfully exported to {queue_filename} (ignored by git by design).")

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(['Baseline', 'Random Forest'], [0.24, 0.74], color=['#cccccc', '#0055ff'])
ax.set_title("Precision@50: Baseline vs. Model")
ax.set_ylabel("Precision Score")

figure_filename = '../../work/figures/precision_vs_baseline.png'
fig.savefig(figure_filename, bbox_inches='tight')
plt.close(fig)
print(f"✅ Figures successfully exported to {figure_filename} for the capstone paper.")
```
